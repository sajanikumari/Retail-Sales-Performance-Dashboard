# retail_data_generator

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
np.random.seed(42)
num_rows = 5150
```

```python
start_date = datetime(2025, 1, 1)
date_list = [start_date + timedelta(days=int(np.random.randint(0, 365))) for _ in range(num_rows)]
```

```python
products = ['Smart Fitness Watch', 'Winter Jacket Waterproof', 'Air Purifier HEPA', 'Designer Handbag', 'Carbon Tennis Racket', 'Casual Sneakers', 'Wireless Headphones']
regions = ['West', 'South', 'Northeast', 'Midwest']
channels = ['In-Store', 'Mobile App', 'Website']
customers = ['Steven Jones', 'Kimberly Hernandez', 'Donna Wilson', 'Anthony Gonzalez', 'Jennifer Brown', 'Michael Smith', 'David Miller']
```

```python
data = {
    'Transaction_ID': [f"TXN_{10000 + i}" for i in range(num_rows)],
    'Date': date_list,
    'Customer_Name': np.random.choice(customers, num_rows),
    'Product_Name': np.random.choice(products, num_rows, p=[0.25, 0.20, 0.15, 0.15, 0.10, 0.10, 0.05]), 
    'Quantity': np.random.choice([1, 2, 3, 4], num_rows, p=[0.5, 0.3, 0.15, 0.05]),
    'Price_Per_Unit': np.random.uniform(50, 500, num_rows).round(2),
    'Cost_Per_Unit': np.random.uniform(25, 250, num_rows).round(2), # Set synthetic cost boundaries for simulation base
    'Region': np.random.choice(regions, num_rows),
    'Sales_Channel': np.random.choice(channels, num_rows, p=[0.16, 0.34, 0.50]) # Define distribution channel allocation probability matrix
}
```

```python
df = pd.DataFrame(data)
```

```python
df['Total_Revenue'] = (df['Quantity'] * df['Price_Per_Unit']).round(2)
df['Total_Cost'] = (df['Quantity'] * df['Cost_Per_Unit']).round(2)
df['Total_Profit'] = (df['Total_Revenue'] - df['Total_Cost']).round(2)
```

```python
df.loc[df['Total_Profit'] < 0, 'Total_Profit'] = df['Total_Revenue'] * 0.4
```

```python
duplicates = df.sample(n=50, random_state=42)
df = pd.concat([df, duplicates], ignore_index=True)
```

```python
output_filename = "retail_sales_data.csv"
df.to_csv(output_filename, index=False)
```

```python
print(f"Tailored CSV '{output_filename}' created with {df.shape[0]} rows!")
```
